```python
import requests
import pandas as pd, os,time,json,sys
sys.path.append("../src")
from coingecko_api import fetch_bitcoin_data
from fear_greed_api import fetch_fear_greed_data
from sentiment_scraper import fetch_news_headlines , fetch_cryptocompare_headlines
```

```python
def save_raw_data(df , filename):
    os.makedirs("../data/raw", exist_ok=True)
    output_path=os.path.join("../data/raw",filename)
    df.to_csv(output_path,index=False)
    print(f"\nSuccess! Cleaned data saved to: {output_path}")



df_bitcoin_price=fetch_bitcoin_data()
df_fear_greed=fetch_fear_greed_data()
df_news_headlines=fetch_news_headlines()
df_cryptocompare_headlines=fetch_cryptocompare_headlines()
df_headlines = pd.concat([df_news_headlines, df_cryptocompare_headlines], ignore_index=True)

# SAVE DATA

save_raw_data(df_bitcoin_price , "bitcoin_price_raw.csv")
save_raw_data(df_fear_greed , "fear_greed_raw.csv")
save_raw_data(df_headlines, "headlines_raw.csv")



    
```
