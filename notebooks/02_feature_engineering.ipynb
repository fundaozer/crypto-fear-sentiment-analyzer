```python
import sys,os
sys.path.append("../src")
import pandas as pd
from feature_engineering import(
    calculate_sentiment_score ,
    merge_dataframes ,
    calculate_panic_score ,
    calculate_features ,
    calculate_market_stress_index
)

df_bitcoin_price=pd.read_csv("../data/raw/bitcoin_price_raw.csv")
df_fear_greed=pd.read_csv("../data/raw/fear_greed_raw.csv")
df_headlines=pd.read_csv("../data/raw/headlines_raw.csv")
```

```python
df_sentiment=calculate_sentiment_score(df_headlines)
df_merged=merge_dataframes(df_bitcoin_price,df_fear_greed,df_sentiment)
df=calculate_features(df_merged)
df=calculate_market_stress_index(df)
df=calculate_panic_score(df)
```

```python
# SAVE THE DATA

os.makedirs("../data/processed" , exist_ok=True)
df.to_csv("../data/processed/bitcoin_processed.csv" , index=False)
print("Success! Processed data saved")
```
